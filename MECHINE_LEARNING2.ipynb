```python

salaries = [
    18, 20, 22, 22, 24,
    25, 25, 26, 28, 30,
    32, 35, 38, 42, 150
]


def calculate_mean(data):
    total = 0
    for value in data:
        total += value
    return total / len(data)



def calculate_median(data):
    sorted_data = sorted(data)
    n = len(sorted_data)
    middle = n // 2

    if n % 2 == 1:
        return sorted_data[middle]
    else:
        return (sorted_data[middle - 1] +
                sorted_data[middle]) / 2



def calculate_modes(data):
    frequencies = {}

    for value in data:
        if value in frequencies:
            frequencies[value] += 1
        else:
            frequencies[value] = 1

    highest_frequency = max(frequencies.values())

    
    if highest_frequency == 1:
        return []

    modes = []
    for value, frequency in frequencies.items():
        if frequency == highest_frequency:
            modes.append(value)

    return modes


print("Mean:", calculate_mean(salaries), "thousand rupees")
print("Median:", calculate_median(salaries), "thousand rupees")
print("Mode(s):", calculate_modes(salaries))
```

```python

data_1 = [12, 15, 15, 18, 20, 22, 25]
data_2 = [5, 5, 10, 10, 15, 20]
data_3 = [7, 14, 21, 28, 35]
data_4 = [100, 100, 100, 100]
data_5 = [1, 2, 3, 4, 5, 100]


def calculate_mean(data):
    if len(data) == 0:
        return None

    total = 0
    for number in data:
        total += number

    return total / len(data)



def calculate_median(data):
    if len(data) == 0:
        return None

    sorted_data = sorted(data)
    n = len(sorted_data)
    middle = n // 2

    if n % 2 == 1:
        return sorted_data[middle]
    else:
        return (sorted_data[middle - 1] + sorted_data[middle]) / 2


def calculate_mode(data):
    if len(data) == 0:
        return []

    counts = {}

    for number in data:
        if number in counts:
            counts[number] += 1
        else:
            counts[number] = 1

    highest_frequency = max(counts.values())

   
    if highest_frequency == 1:
        return []

    modes = []
    for number, frequency in counts.items():
        if frequency == highest_frequency:
            modes.append(number)

    return sorted(modes)


print("Mean:", calculate_mean(data_1))
print("Median:", calculate_median(data_1))
print("Mode:", calculate_mode(data_1))
```

```python

import numpy as np
import pandas as pd

scores = [45, 50, 52, 55, 58, 60, 65, 75]


total = 0
for score in scores:
    total += score

n = len(scores)
mean = total / n

squared_deviations = []

for score in scores:
    squared_deviations.append((score - mean) ** 2)

sum_squared_deviations = sum(squared_deviations)


population_variance = sum_squared_deviations / n
population_std = population_variance ** 0.5


sample_variance = sum_squared_deviations / (n - 1)
sample_std = sample_variance ** 0.5


results = pd.DataFrame({
    "Measure": [
        "Number of employees",
        "Mean",
        "Population variance",
        "Sample variance",
        "Population standard deviation",
        "Sample standard deviation"
    ],
    "Value": [
        n,
        mean,
        population_variance,
        sample_variance,
        population_std,
        sample_std
    ]
})

print(results.round(4).to_string(index=False))
```

```python

import seaborn as sns
import pandas as pd


df = sns.load_dataset("tips")

def population_stats(data):
    values = data.dropna().tolist()
    n = len(values)

    mean = sum(values) / n

    squared_deviations = []
    for value in values:
        squared_deviations.append((value - mean) ** 2)

    variance = sum(squared_deviations) / n
    std_dev = variance ** 0.5
    cv = (std_dev / mean) * 100 if mean != 0 else None

    return {
        "Count": n,
        "Mean": mean,
        "Population Variance": variance,
        "Population Std Dev": std_dev,
        "CV (%)": cv
    }


results = pd.DataFrame({
    "total_bill": population_stats(df["total_bill"]),
    "tip": population_stats(df["tip"])
}).T

print(results.round(3))
```

```python

import pandas as pd

# Four datasets
data_1 = [10, 20, 30, 40, 50]
data_2 = [20, 20, 20, 20, 20]
data_3 = [5, 10, 15, 20, 100]
data_4 = [12, 14, 16, 18, 20, 22]


def calculate_statistics(data):
    n = len(data)

    if n == 0:
        raise ValueError("Dataset cannot be empty.")

   
    mean = sum(data) / n

   
    squared_deviations = []
    for value in data:
        squared_deviations.append((value - mean) ** 2)

    total_squared_deviations = sum(squared_deviations)

  
    population_variance = total_squared_deviations / n
    population_std = population_variance ** 0.5

    
    if n > 1:
        sample_variance = total_squared_deviations / (n - 1)
        sample_std = sample_variance ** 0.5
    else:
        sample_variance = None
        sample_std = None

    return {
        "Count": n,
        "Mean": mean,
        "Population Variance": population_variance,
        "Sample Variance": sample_variance,
        "Population Std Dev": population_std,
        "Sample Std Dev": sample_std
    }



datasets = {
    "Data 1": data_1,
    "Data 2": data_2,
    "Data 3": data_3,
    "Data 4": data_4
}


results = {}

for name, data in datasets.items():
    results[name] = calculate_statistics(data)

report = pd.DataFrame(results).T

print("VARIANCE AND STANDARD DEVIATION REPORT")
print(report.round(2).to_string())
```

```python

measurements = [12, 15, 18, 20, 25, 30, 40]


def calculate_stats(data):
    n = len(data)
    mean = sum(data) / n

    squared_deviations = [(x - mean) ** 2 for x in data]
    variance = sum(squared_deviations) / n
    std_dev = variance ** 0.5

    return mean, variance, std_dev



plus_10 = [x + 10 for x in measurements]
times_4 = [x * 4 for x in measurements]
times_minus_2 = [x * -2 for x in measurements]


original = calculate_stats(measurements)
added = calculate_stats(plus_10)
multiplied = calculate_stats(times_4)
negative = calculate_stats(times_minus_2)

print("Add 10:")
print("Mean increases by 10:",
      abs(added[0] - (original[0] + 10)) < 1e-9)
print("Variance unchanged:",
      abs(added[1] - original[1]) < 1e-9)
print("Standard deviation unchanged:",
      abs(added[2] - original[2]) < 1e-9)

print("\nMultiply by 4:")
print("Mean multiplied by 4:",
      abs(multiplied[0] - original[0] * 4) < 1e-9)
print("Variance multiplied by 16:",
      abs(multiplied[1] - original[1] * 16) < 1e-9)
print("Standard deviation multiplied by 4:",
      abs(multiplied[2] - original[2] * 4) < 1e-9)

print("\nMultiply by -2:")
print("Mean multiplied by -2:",
      abs(negative[0] - original[0] * -2) < 1e-9)
print("Variance multiplied by 4:",
      abs(negative[1] - original[1] * 4) < 1e-9)
print("Standard deviation multiplied by 2:",
      abs(negative[2] - original[2] * 2) < 1e-9)

```

```python

sales = [12, 14, 15, 15, 16, 17, 18,
         19, 20, 21, 22, 24, 25, 120]


def calculate_stats(data):
    n = len(data)
    mean = sum(data) / n

    sorted_data = sorted(data)
    middle = n // 2

    if n % 2 == 0:
        median = (sorted_data[middle - 1] +
                  sorted_data[middle]) / 2
    else:
        median = sorted_data[middle]

    squared_deviations = [(x - mean) ** 2 for x in data]
    variance = sum(squared_deviations) / n
    std_dev = variance ** 0.5

    return mean, median, variance, std_dev



sorted_sales = sorted(sales)
n = len(sorted_sales)

lower_half = sorted_sales[:n // 2]
upper_half = sorted_sales[n // 2:]

def find_median(data):
    length = len(data)
    middle = length // 2

    if length % 2 == 0:
        return (data[middle - 1] + data[middle]) / 2
    else:
        return data[middle]

Q1 = find_median(lower_half)
Q3 = find_median(upper_half)

IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = [x for x in sales
            if x < lower_bound or x > upper_bound]


mean, median, variance, std_dev = calculate_stats(sales)

print("Mean:", round(mean, 2))
print("Median:", round(median, 2))
print("Population variance:", round(variance, 2))
print("Population standard deviation:", round(std_dev, 2))

print("\nQ1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Flagged outliers:", outliers)
```
